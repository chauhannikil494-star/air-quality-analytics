### Handling Missing Values in Numerical Columns

Based on the missing-value analysis and the distribution of the numerical
columns, we will use the median to fill missing values in the numerical
pollutant columns.

The median is preferred because several numerical columns contain
relatively high or extreme values. Unlike the mean, the median is less
affected by extreme observations.

The `AQI` column will be handled separately because it represents the
overall Air Quality Index and may be used as an important target or
analysis variable later in the project.

```python
numerical_columns = df.select_dtypes(include="number").columns.drop("AQI")

df[numerical_columns] = df[numerical_columns].fillna(
    df[numerical_columns].median()
)
```

```python
df[numerical_columns].isnull().sum()
```

### Numerical Missing Values Cleaning Result

The missing values in all numerical pollutant columns have been
successfully handled using the median value of their respective columns.

After the cleaning process, all selected numerical pollutant columns
contain zero missing values.

The `AQI` column was intentionally excluded from this step and will be
handled separately because it represents the overall Air Quality Index.

### Relationship Between Pollutants and AQI

Air Quality Index (AQI) is influenced by air pollution measurements.
Therefore, it is useful to examine the relationship between individual
pollutants and AQI.

In this section, we will calculate the correlation between major
pollutants and AQI.

Correlation measures the strength and direction of a linear relationship
between two numerical variables.

A positive correlation indicates that the two variables tend to increase
together, while a negative correlation indicates that one tends to
decrease as the other increases.

Correlation does not prove that one variable causes changes in another.

```python
pollutants = ["PM2.5", "PM10", "NO2", "CO", "SO2", "O3"]

aqi_correlation = df_aqi[pollutants + ["AQI"]].corr()["AQI"].drop("AQI")

aqi_correlation.sort_values(ascending=False)
```

### Pollutant-AQI Correlation — Interpretation

The correlation analysis shows a positive correlation between all six
selected pollutants and AQI in this dataset.

The calculated correlation values are:

- CO: 0.678071
- PM2.5: 0.655945
- NO2: 0.535813
- SO2: 0.485029
- PM10: 0.456212
- O3: 0.194327

The results indicate that the selected pollutants have different levels
of linear association with AQI in the analyzed data.

Among the selected pollutants, CO and PM2.5 have the largest correlation
values with AQI, while O3 has the smallest correlation value.

These results describe statistical association within this dataset and
should not be interpreted as proof that a pollutant directly causes
changes in AQI.

### Correlation Heatmap

A correlation heatmap provides a visual representation of the correlation
between numerical variables.

In this analysis, we will visualize the correlations among the selected
pollutants and AQI.

The heatmap makes it easier to identify positive and negative linear

```python
correlation_data = df_aqi[pollutants + ["AQI"]].corr()

plt.figure(figsize=(10, 7))

sns.heatmap(correlation_data, annot=True, cmap="coolwarm", fmt=".2f")

plt.title("Correlation Heatmap of Pollutants and AQI")
plt.tight_layout()
plt.show()
```

### Correlation Heatmap — Interpretation

The correlation heatmap provides a visual representation of the

The heatmap confirms that the selected pollutants have positive

The correlation values observed earlier show relatively stronger positive

The heatmap is useful for quickly identifying the direction and relative

### Monthly Average AQI — Interpretation

The monthly AQI trend shows that the average AQI varies across different
months of the year.

The line chart helps identify months in which the average AQI is
relatively higher or lower.

throughout the year.

available cities and observations. It does not show the monthly trend of

seasonal changes in air quality.

## Season-wise AQI Analysis

### Understanding Seasonal AQI Patterns

Air quality can vary across different seasons due to changes in weather
conditions and atmospheric factors.

This analysis groups the AQI observations into four broad seasons and
compares their average AQI values.

The comparison helps identify seasonal patterns in air quality within the
dataset.

```python
def get_season(month):
    if month in [12, 1, 2]:
        return "Winter"
    elif month in [3, 4, 5]:
        return "Summer"
    elif month in [6, 7, 8, 9]:
        return "Monsoon"
    else:
        return "Post-Monsoon"


air_quality["Season"] = air_quality["Date"].dt.month.apply(get_season)
```

```python
seasonal_aqi = (
    air_quality.groupby("Season")["AQI"]
    .mean()
)

seasonal_aqi
```

```python
plt.figure(figsize=(10, 6))

seasonal_aqi.plot(kind="bar")

plt.title("Season-wise Average AQI")
plt.xlabel("Season")
plt.ylabel("Average AQI")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

### Season-wise AQI Analysis — Interpretation

The bar chart compares the average AQI across the four seasons defined in this
analysis.

The average AQI differs across seasons, indicating that air quality conditions
vary during different parts of the year.

Winter shows a relatively higher average AQI, while the Monsoon season shows a
relatively lower average AQI.

Summer and Post-Monsoon fall between these two seasonal levels.

Overall, the analysis indicates a seasonal variation in AQI within the dataset.
However, these results describe the observations available in this dataset and
do not by themselves establish the causes of the seasonal differences.
